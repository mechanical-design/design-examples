The clamp assembly as shown in Figure Q1 consists of member AB and AC, which are pin connected at A and the compressive stress on the wood between B and C is 450 kPa. Determine;

1. the torque required to tighten the screw, 

2. the power if the screw travels at a speed of 2 mm/s,

3. the normal force if the length handle DE is reduced to 40 mm, and

4. whether the screw will overhaul if a ball thrust bearing (negligible fc) were used in    place of the plain thrust collar

Design assumption: The clamp is a single ACME thread of M12 x 2. The plain thrust collar diameter, dc is 13.5 mm and coefficient of friction between all contact surfaces is 0.3.


```python
sigma = 450e3
screw_type = 1 # single-start
alpha = 14.5
d_outer = 12/1000
pitch = 2/1000
f_screw = 0.3
f_collar = 0.3
d_collar = 13.5/1000
```

1. Torque required to tighten the screw

```python
d_inner= (d_outer - pitch)
print(f'Root diameter d_inner = {d_inner} m')
print(f'Root diameter d_inner = {d_inner*1000} mm')
```

Stress,
$$\sigma = \dfrac{Load}{\text{Bolt cross sectional area}}$$

Use dr to find area for power screw

$$450 \times 10^3 = \dfrac{Load}{\dfrac{\pi (0.01)^2}{4}}$$

```python
Fc = sigma * (pi/4) * d_inner**2
print(f'Force at C = {Fc:.4e} N')
```

Draw FBD for link AC. Find load at power screw,W

$$Fc(0.085)-W(0.04)=0$$

```python
d_ac = 0.085
d_ae = 0.04
Load = Fc*d_ae/d_ac
print(f'Load = {Load:.4e} N')
```

```python
d_mean = (d_outer+d_inner)/2

print(f'Mean diameter={d_mean}mm')
```

```python
Lead = screw_type*pitch
print(f'Lead={Lead}m')
print(f'Lead={Lead*1000}mm')

tan_lambda = Lead/(pi*d_mean)
print(f'tan(lambda)={tan_lambda}')
lambda_helix = round(degrees(atan(tan_lambda)),3)
print(f'Lambda = {lambda_helix}')
```

```python
alpha_n = round(degrees(atan(cos(radians(lambda_helix))*tan(radians(alpha)))),2)
print(f'alpha_n = {alpha_n}')
```

```python
T_u =(Load*d_mean/2)*((f_screw+(cos(radians(alpha_n))*tan(radians(lambda_helix))))/(cos(radians(alpha_n))-(f_screw*tan(radians(lambda_helix)))))+(Load*f_collar*d_collar/2)
print(f'T_u={T_u:.2e}N')
```

2. Power

Given v = 2mm/s

$$v_{screw}=Lead \times \dfrac{n_{rpm}}{60}$$

$$2=(2) \times \dfrac{n_{rpm}}{60}$$

```python
v_scrw = 2/1000 #m/s
n_rpm = (v_scrw/Lead)*60
print(f'n = {n_rpm:.2f} rpm')
```

```python
Power = (2*pi*n_rpm/60)*T_u
print(f'Power = {Power:.2e} W')
```

3. Normal force

$$T=F \times d$$

```python
d_de=0.04
F_nominal = T_u/d_de
print(f'Nominal Force = {F_nominal} N')
```

4. Overhaul

$$f > \cos \alpha_n \tan \lambda$$

```python
overhaul = cos(radians(alpha_n))*tan(radians(lambda_helix))
print(f'Overhaul check value = {overhaul}')
```

The power screw is in self-locking condition because $f > \cos \alpha_n \tan \lambda$
