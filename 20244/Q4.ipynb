b)

```python
N1=50
N2=125
N3=25
N4=75
module=4
pressureAngle=radians(20)
```

i)

$$\frac{n_2}{n_1}=\frac{N_1}{N_2}=\frac{50}{125}=\frac{2}{5}$$

$$\frac{n_4}{n_3}=\frac{N_3}{N_4}=\frac{25}{75}=\frac{1}{3}$$

$$\frac{n_4}{n_1}=\frac{n_4}{n_3}\times\frac{n_2}{n_1}$$

```python
n2_n1=N1/N2
n4_n3=N3/N4
gear_ratio=n2_n1*n4_n3
print(f'Total gear ratio = {Fraction(gear_ratio).limit_denominator()}')
```

ii)

```python
Power = 40000
n1=1500
T1 = (60*Power)/(2*pi*n1)
print(f'Torque on gear 1: {T1:.2f} Nm')
```

$$module=\frac{d_1}{N_1}$$

```python
d1 = module * N1/1000
print(f'Diameter of gear 1: {d1*1000:.2f} mm')
r1 = d1 / 2
print(f'Radius of gear 1: {r1*1000:.2f} mm')
```

$$Torque=F_{t1}r_1$$

```python
Ft1=T1/r1
print(f'Tangential force on gear 1: {Ft1:.2f} N')
Ft2=Ft1
print(f'Tangential force on gear 2: {Ft2:.2f} N')
```

$$F_{r1}=F_{t1}\tan(20)$$

```python
Fr1=Ft1*tan(pressureAngle)
print(f'Radial force on gear 1: {Fr1:.2f} N')
Fr2=Fr1
print(f'Radial force on gear 2: {Fr2:.2f} N')
```

$$\frac{n_2}{n_1}=\frac{N_1}{N_2}$$

```python
n2= (N1/N2)*1500
print(f'Speed of gear 2: {n2:.2f} rpm')
```

```python
T2 = (60*Power)/(2*pi*n2)
print(f'Torque on gear 2: {T2:.2f} Nm')
```

```python
d3 = module * N3/1000
print(f'Diameter of gear 3: {d3*1000:.2f} mm')
r3 = d3 / 2
print(f'Radius of gear 3: {r3*1000:.2f} mm')
```

```python
Ft3=T2/r3
print(f'Tangential force on gear 3: {Ft3:.2f} N')
Ft4=Ft3
print(f'Tangential force on gear 4: {Ft4:.2f} N')
```

```python
Fr3=Ft3*tan(pressureAngle)
print(f'Radial force on gear 3: {Fr3:.2f} N')
Fr4=Fr3
print(f'Radial force on gear 4: {Fr4:.2f} N')
```
