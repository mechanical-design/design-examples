```python
Pload=46000
Nbolt=6
FOS=2.6
Sp=650e6
print(f'Class 8.8 bolt proof strength: {Sp/1e6} MPa')
# Eb=2*Ep
# Ap =3*Ab
```

$$C = \frac{k_b}{k_b + K_p}$$
$$ = \frac{\frac{A_b E_b}{L}}{\frac{A_b E_b}{L} + \frac{A_p E_p}{L}} $$
$$= \frac{\frac{A_b 2 E_p}{L}}{\frac{A_b 2 E_p}{L} + \frac{3 A_b E_p}{L}} $$
$$= \frac{2}{5} = 0.4$$

```python
C=0.4
```

$$S_p A_t= C\left(\frac{P}{N}\right)n + F_i$$
$$650\times 10^6 A_t= 0.4\left(\frac{46\times 10^3}{6}\right)2.6 + 0.75 (650\times 10^6 A_t)$$

```python
A_t=C*(Pload/Nbolt)*FOS/(Sp*(1-0.75))
print(f'A_t: {A_t:.3e} m^2')
print(f'A_t: {A_t*1e6:.2f} mm^2')
print(f'From Table At= 58 mm^2 with d=10mm')
```

ii)

```python
Nbolt=4
Fbolt=C*(Pload/Nbolt)*FOS+(0.75*Sp*A_t)
print(f'Fbolt: {Fbolt:.2f} N')
```

```python
print(f'External load per bolt: {Pload/Nbolt:.2f} N')
```

```python
print(f'{Fbolt:.2f}N > {Pload/Nbolt:.2f}N, thus the design is not separated.')
```
