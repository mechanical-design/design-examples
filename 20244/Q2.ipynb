```python
d_input=0.3
n_input = 2400
d_output=0.5
c_distance=576e-3
friction = 0.25
density = 1000  #kg/m3
sigma = 2.8e6
F2=250
```

$$\sigma = \frac{F_1}{A}$$
$$F_1=2.8\times 10^6 A$$

$$\frac{F_1 - F_c}{F_2 - F_c} = e^{f\phi}$$
$$F_c=\rho A V^2$$
$$V = \frac{\pi d n_{rpm}}{60} $$
$$= \pi(0.3)\left(\frac{2400}{60}\right) = 37.67$$

$$F_c=1000A(37.67)^2=1.42\times 10^6$$


$$\frac{2.8 \times 10^6 A - 1.42 \times 10^6 A}{250 - 1.42 \times 10^6 A} = e^{0.25(2.79)}$$

$$A = 119.6\times 10^{-6}m^2$$

```python
A=119.6e-6
print(f'Cross-sectional area A: {A:.4e} m^2')
print(f'Cross-sectional area A: {A*1e6:.2f} mm^2')
```

```python
print(f'Power transmitted')
```

$$P=\frac{2\pi n_{rpm}}{60}T$$
where
$$T = (F_1 - F_2)r_1$$
$$F_1=2.8\times 10^6 A$$

```python
F1=sigma*A
print(f'Allowable load F1: {F1} N')
```

```python
r_input=d_input/2
Torque=(F1-F2)*r_input
print(f'Torque transmitted: {Torque} Nm')
```

```python
Power=2*pi*n_input*Torque/60
print(f'Power transmitted: {Power:.2f} W')
```
