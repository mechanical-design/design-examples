Draw SFD and BMD for Vertical

First, find reaction force

```python
# Vertical Loading

coorA = [0,0]
coorB = [320,0]
coorC = [120,0]
coorD = [370,0]

data = np.array([coorA,coorB,coorC,coorD])
x, y = data.T

fig, ax = plt.subplots()
fig.set_size_inches(6,2)
ax.grid()
ax.set_yticks([0])
ax.scatter(x,y, color='k')

ax.arrow(coorA[0],coorA[1], 0,120,width=5,color='r')
ax.arrow(coorC[0],coorC[1]-150, 0,120,width=5,color='r')
ax.arrow(coorD[0],coorD[1]-150, 0,120,width=5,color='r')
ax.hlines(0,0,370, colors='k')

ax.set(xlabel='l(mm)', ylabel='y',
       title='FBD for Shaft AD (Vertical)')

ax.text(coorA[0]+2,coorA[1]+130,f'Ay={Fa} sin60')
# ax.text(coorB[0]+2,coorB[1]-30,f'By')
# ax.text(coorA[0]+2,coorA[1]+130,f'Ay')
ax.text(coorC[0]+2,coorC[1]-130,f'Cy')
ax.text(coorD[0]+2,coorD[1]-130,f'Dy')
ax.text(coorA[0]-50,coorA[1]+5,f'  A 120mm')
ax.text(coorB[0],coorB[1]+5,f'  B 50mm')
ax.text(coorC[0],coorC[1]+5,f' C 200mm')
ax.text(coorD[0],coorD[1]-20,f' D')

plt.axis('equal')
plt.show()
```

$$+\circlearrowleft \sum M_d=0$$
$$-Ay(0.37)-Cy(0.25)=0$$

```python
Ay=Fa*sin(radians(60))
print(f'Reaction at A in vertical direction is {Ay:.2f} N')
```

```python
Cy=-Ay*0.37/0.25
print(f'Reaction at C in vertical direction is {Cy:.2f} N')
```

$$+\uparrow \sum F_y=0$$
$$Cy+Dy+Ay=0$$

```python
Dy=-Ay-Cy
print(f'Reaction at D in vertical direction is {Dy:.2f} N')
```

```python
ac=120
ab=320
ad=370

y=[0,0,ac,ac,ad,ad]
Vy=[0,Ay,Ay,Ay+Cy,Ay+Cy,0]

fig, ax1 = plt.subplots()
ax1.set_yticks([0])
ax1.plot(y,Vy)
ax1.hlines(0,0,ad,colors='black')
ax1.fill_between(y, Vy, alpha=0.3) # Added fill_between
ax1.set_title('Shear Force Diagram (SFD)')
ax1.set_xlabel('l (mm)')
ax1.set_ylabel('Shear Force,V (kN)')

ax1.text(coorA[0],coorA[1]+100,f'  A')
ax1.text(coorB[0],coorB[1]+100,f' B')
ax1.text(coorC[0],coorC[1]+100,f'C')
ax1.text(coorD[0],coorD[1]+100,f'D')

ax1.text(coorA[0],Vy[1]+100,f'        {Vy[1]:.2f}N')
ax1.text(coorC[0],Vy[3]+100,f'         {Vy[3]:.2f}N')

plt.show()
```

```python
ac=120
ab=320
bc=300
ad=370

y=[0,ac,ab,ad]
Mcy=Ay*ac/1000
Mby=(Dy*bc/1000)-Mcy
My=[0,Mcy,Mby,0]

fig, ax2 = plt.subplots()
ax2.set_yticks([0])
fig.set_size_inches(6,3)
ax2.plot(y, My, marker='x')
ax2.hlines(0,0,ad,colors='black')
ax2.fill_between(y, My, alpha=0.3) # Added fill_between
ax2.set_title('Bending MomentDiagram (BMD)')
ax2.set_xlabel('l (mm)')
ax2.set_ylabel('Bending Moment,M (kNm)')
ax2.text(coorA[0],coorA[1]+5,f'A')
ax2.text(coorB[0],coorB[1]+5,f'B')
ax2.text(coorC[0],coorC[1]+5,f'C')
ax2.text(coorD[0],coorD[1]+5,f'D')
ax2.text(coorB[0],My[2],f'Mby={My[2]:.2f}Nm')
ax2.text(coorC[0],My[1],f'Mcy={My[1]:.2f}Nm')
plt.tight_layout()
plt.show()
```
