1. Joint Stiffness constant,C

```python
from math import pi, log, atan, degrees, radians, cos, tan, sqrt
```

```python
Pressure = 200e3
FOS=1.5
print('Joint is reusable')
# Ep=0.5*Eb
d=0.02
L=0.05
```

```python
At=245e-6
Sp=600e6
Fi=Sp*At
print(f'Fi={Fi/1e3:.1f} kN')
```

```python
Ab=pi*(d/2)**2
kb=Ab/L
print(f'Bolt stiffness factor, kb={kb:.3e} Es')
```

```python
kp=(0.58*pi*d/2)/(2*log(5*(((0.58*L)+(0.5*d))/((0.58*L)+(2.5*d)))))
print(f'Part stiffness factor, kp={kp:.3e} Es')
```

```python
C = round(kb/(kb+kp),4)
# print('C=', f"{C:.3e}")
print(f'C= {C}')
```

2. No of bolt

$$S_p A_t = \dfrac {CP(1.5)}{Nbolt}+0.75S_p A_t$$

```python
print('Find load P from pressure and area')
d_flange=0.25*2
Area=pi*(d_flange/2)**2
P=Pressure*Area
print(f'P={P/1e3:.1f} kN')
```

```python
Nbolt=C*P*FOS/(0.25*Sp*At)
print(P, Sp, At, C, FOS)
print(f'Number of bolts, Nbolt={Nbolt:.2f}')
```
